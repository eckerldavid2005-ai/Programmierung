```python
import pandas as pd
import numpy as np
import pathlib
import matplotlib.pyplot as plt
import sklearn
datapath = pathlib.Path("KI04_Praktikum_pythons") / "datasets"
```

```python
gdppc_col = "GDP per capita (USD)"
lifesat_col = "Life satisfaction"
full_country_stats = pd.read_csv(datapath / "lifesat_full.csv")
```

```python
x_sample = full_country_stats[[gdppc_col]].values
y_sample = full_country_stats[lifesat_col].values
print(x_sample[:5])
print(y_sample[:5])
```

```python
from sklearn.linear_model import LinearRegression

# Messpunkte anzeigen
plt.scatter(x_sample, y_sample)

# Modell erstellen
lin1 = LinearRegression()

# Modell trainieren
lin1.fit(x_sample, y_sample)

# Werte für die Regressionsgerade erzeugen
new_x = np.linspace(min(x_sample), max(x_sample), 100)

# Vorhersage berechnen
new_y = lin1.predict(new_x.reshape((-1, 1)))
```

```python
plt.scatter(x_sample, y_sample)
plt.plot(new_x, new_y, color="red")
plt.show()
```
